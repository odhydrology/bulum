# QAStat2s Report Card Demo

This notebook demonstrates how to generate hydrology report cards with
`bulum.stats.qastat2s`, comparing modelled flows against observed data.

> The CSV paths below point at the package's bundled sample data
> (`tests/`). Adjust them to your own files.

## Setup

Run this cell first to enable high-resolution display:

```python
# Enable high-resolution display for crisp figures
%config InlineBackend.figure_format = 'retina'
%config InlineBackend.figure_dpi = 300
```

## Option 1: From CSV Files

Use `report_card()` when your data is in CSV files:

```python
import bulum.stats.qastat2s as qastat2s

fig = qastat2s.report_card(
    obs_file="tests/146012A.csv",
    mod_file="tests/result_flow.csv",
    rain_file="tests/058067.infilled.csv",
    heading="Report Card from CSV Files"
)
```

## Option 2: From in-memory data

Use `report_card_from_data()` when your data is already loaded. Inputs can be
pandas Series, or single-column bulum `TimeseriesDataframe` / DataFrame columns
(the native bulum types returned by `bulum.io.read_ts_csv`):

```python
from bulum import io
import bulum.stats.qastat2s as qastat2s

# Read with bulum.io into TimeseriesDataframes (allow_nonnumeric tolerates a
# quality/flag column alongside the flow column).
obs = io.read_ts_csv("tests/146012A.csv", allow_nonnumeric=True)
mod = io.read_ts_csv("tests/result_flow.csv", allow_nonnumeric=True)
rain = io.read_ts_csv("tests/058067.infilled.csv", allow_nonnumeric=True)

# Pass single columns (Series or single-column TSDF both work)
fig = qastat2s.report_card_from_data(
    obs=obs.iloc[:, 0],
    mod=mod.iloc[:, 0],
    rain=rain.iloc[:, 0],
    heading="Report Card from in-memory data"
)
```

## Custom Date Range

Both functions support custom analysis periods:

```python
fig = qastat2s.report_card(
    obs_file="tests/146012A.csv",
    mod_file="tests/result_flow.csv",
    rain_file="tests/058067.infilled.csv",
    heading="Custom Date Range",
    start_date="1980-01-01",
    end_date="2000-12-31"
)
```

## Custom Water Year

The `water_year_start_month` parameter controls annual aggregation (default 7 = July for Australian convention):

```python
# Generate report card from in-memory data with an October water year
fig = qastat2s.report_card_from_data(
    obs=obs.iloc[:, 0],
    mod=mod.iloc[:, 0],
    rain=rain.iloc[:, 0],
    heading="October Water Year",
    water_year_start_month=10
)
```

## Custom Colours

Use `get_colors()` to see and modify the colour palette:

```python
colors = qastat2s.get_colors()
print("Available colour keys:", list(colors.keys()))

# Customize colours
colors['observed'] = '#1f77b4'  # Blue
colors['modelled'] = '#ff7f0e'  # Orange

fig = qastat2s.report_card(
    obs_file="tests/146012A.csv",
    mod_file="tests/result_flow.csv",
    rain_file="tests/058067.infilled.csv",
    heading="Custom Colours",
    colors=colors
)
```

## Function Reference

All functions are accessed via `import bulum.stats.qastat2s as qastat2s`.

### `qastat2s.report_card()` - From CSV files

| Parameter | Type | Description |
|-----------|------|-------------|
| `obs_file` | str | Path to observed flow CSV |
| `mod_file` | str | Path to modelled flow CSV |
| `rain_file` | str | Path to rainfall CSV (optional) |
| `heading` | str | Title for the report card |
| `start_date` | str | Analysis start date (optional) |
| `end_date` | str | Analysis end date (optional) |
| `p_low` | float | Low flow threshold (default 0.79) |
| `p_high` | float | High flow threshold (default 0.1) |
| `water_year_start_month` | int | Month when water year starts, 1-12 (default 7 = July) |
| `colors` | dict | Custom color palette from `get_colors()` (optional) |
| `output_file` | str | Path to save figure (optional) |
| `dpi` | int | Resolution (default 300) |

Files are read with `bulum.io.read_ts_csv`; the first data column is used.

### `qastat2s.report_card_from_data()` - From in-memory data

| Parameter | Type | Description |
|-----------|------|-------------|
| `obs` | Series / single-col DataFrame / TSDF | Observed flow (ML/d) |
| `mod` | Series / single-col DataFrame / TSDF | Modelled flow (ML/d) |
| `rain` | Series / single-col DataFrame / TSDF | Rainfall (mm/d, optional) |
| `heading` | str | Title for the report card |
| `start_date` | str | Analysis start date (optional) |
| `end_date` | str | Analysis end date (optional) |
| `p_low` | float | Low flow threshold (default 0.79) |
| `p_high` | float | High flow threshold (default 0.1) |
| `water_year_start_month` | int | Month when water year starts, 1-12 (default 7 = July) |
| `colors` | dict | Custom color palette from `get_colors()` (optional) |
| `output_file` | str | Path to save figure (optional) |
| `dpi` | int | Resolution (default 300) |

Indexes may be datetime-like or bulum-style `'YYYY-MM-DD'` date strings.

### `qastat2s.get_colors()` - Get default color palette

Returns a dict with keys: `primary`, `observed`, `modelled`, `rainfall`, `subtle`, `text_secondary`, `grid`, `white`, `black`
